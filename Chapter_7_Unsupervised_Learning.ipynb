# Chapter 7: Unsupervised Learning

Berbeda dengan *supervised learning* yang bergantung pada label target (variabel $Y$), **unsupervised learning** berfokus pada eksplorasi data tanpa label untuk menemukan pola tersembunyi, struktur alami, atau pengelompokan di dalam data. Bab ini membahas teknik utama seperti *Principal Component Analysis* (PCA), *K-Means Clustering*, dan *Hierarchical Clustering*.

---

## 1. Principal Component Analysis (PCA)
* **Definisi:** Teknik reduksi dimensi linier yang mengubah kumpulan variabel berkorelasi menjadi sekumpulan variabel baru yang tidak berkorelasi yang disebut **Principal Components (Komponen Utama)**.
* **Tujuan:** Mengurangi kompleksitas data dan jumlah dimensi (fitur) sambil mempertahankan sebesar mungkin varians atau informasi asli dari data.
* **Eigenvalues & Variance:** Komponen utama pertama ($PC1$) merangkum varians terbesar dalam data, diikuti oleh $PC2$, $PC3$, dan seterusnya secara ortogonal.

---

## 2. K-Means Clustering
* **Definisi:** Algoritma pengelompokan non-hierarki yang mempartisi $N$ pengamatan ke dalam $K$ kluster, di mana setiap pengamatan milik kluster dengan rata-rata terdekat (*centroid*).
* **Langkah Kerja Algoritma:**
  1. Menentukan jumlah kluster $K$ secara acak atau menggunakan metode *Elbow*.
  2. Menginisialisasi $K$ pusat kluster (*centroids*).
  3. Menetapkan setiap titik data ke *centroid* terdekat (berdasarkan jarak Euclidean).
  4. Memperbarui posisi *centroid* berdasarkan rata-rata titik dalam kluster tersebut.
  5. Mengulangi langkah 3 dan 4 hingga konvergen (posisi *centroid* tidak berubah).

---

## 3. Hierarchical Clustering
* **Definisi:** Metode pengelompokan yang membangun hierarki kluster, baik secara *agglomerative* (pendekatan bawah-ke-atas di mana setiap titik dimulai sebagai kluster terpisah lalu digabungkan secara bertahap) maupun *divisive* (atas-ke-bawah).
* **Dendrogram:** Diagram pohon yang memvisualisasikan proses penggabungan atau pemisahan kluster pada setiap tingkat jarak, membantu analis menentukan jumlah kluster optimal yang terbentuk.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# ==========================================
# 1. Menarik Dataset dari GitHub
# ==========================================
# Menggunakan dataset 'state.csv' dari repositori resmi penulis untuk analisis klaster & PCA
url = "https://raw.githubusercontent.com/gedeck/practical-statistics-for-data-scientists/master/data/state.csv"
state_data = pd.read_csv(url)

print("--- 5 Baris Pertama Dataset State ---")
display(state_data.head())

# ==========================================
# 2. Persiapan & Standarisasi Data
# ==========================================
# Memilih fitur numerik untuk klastering dan reduksi dimensi
X = state_data[['Population', 'Murder.Rate', 'Household.Income']]

# Melakukan standarisasi agar setiap fitur memiliki mean=0 dan standard deviation=1
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# ==========================================
# 3. Model K-Means Clustering
# ==========================================
# Menentukan jumlah kluster K = 3
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
state_data['Cluster'] = kmeans.fit_predict(X_scaled)

print("\n--- Hasil Klasterisasi K-Means (Jumlah per Klaster) ---")
print(state_data['Cluster'].value_counts())

# ==========================================
# 4. Principal Component Analysis (PCA)
# ==========================================
# Mereduksi dimensi fitur menjadi 2 Komponen Utama (PC1 dan PC2)
pca = PCA(n_components=2)
principal_components = pca.fit_transform(X_scaled)

state_data['PC1'] = principal_components[:, 0]
state_data['PC2'] = principal_components[:, 1]

print("\n--- Explained Variance Ratio dari PCA ---")
print(f"PC1: {pca.explained_variance_ratio_[0]*100:.2f}%")
print(f"PC2: {pca.explained_variance_ratio_[1]*100:.2f}%\n")

# ==========================================
# 5. Visualisasi Hasil K-Means & PCA
# ==========================================
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=state_data,
    x='PC1',
    y='PC2',
    hue='Cluster',
    palette='Set1',
    s=100,
    alpha=0.8
)
plt.title('PCA Visualization with K-Means Clusters')
plt.xlabel('Principal Component 1 (PC1)')
plt.ylabel('Principal Component 2 (PC2)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(title='Cluster')
plt.show()
```
